```python
import streamlit as st
import google.generativeai as genai
from PIL import Image

# ==========================================
# CẤU HÌNH BỘ NÃO AI (GEMINI)
# ==========================================
# ⚠️ CẢNH BÁO BẢO MẬT: Không chia sẻ API Key này public
# Trong thực tế, bạn nên dùng biến môi trường (Environment Variables)
GOOGLE_API_KEY = "AQ.Ab8RN6Io6x_G-kg39i5Qpbez3Ijq001GKL3Xx4yItdoDt51UAw"
genai.configure(api_key=GOOGLE_API_KEY)

# Khởi tạo mô hình Gemini 1.5 Flash (chuyên xử lý nhanh ảnh + văn bản)
model = genai.GenerativeModel('gemini-1.5-flash')

# ==========================================
# CẤU HÌNH GIAO DIỆN (STREAMLIT UI)
# ==========================================
st.set_page_config(page_title="App Đọc Chữ Viết Tay", page_icon="✍️")
st.title("✍️ Trợ Lý Đọc Chữ Viết Tay AI")
st.subheader("Tải ảnh chụp ghi chú của bạn lên, AI sẽ đọc giúp bạn.")

# --- Phần 1: Tải ảnh ---
uploaded_file = st.file_uploader("Chọn một hình ảnh chứa chữ viết tay...", type=["jpg", "jpeg", "png"])

if uploaded_file is not None:
    # Hiển thị ảnh đã tải lên
    image = Image.open(uploaded_file)
    st.image(image, caption='Ảnh bạn đã tải lên', use_column_width=True)
    
    # --- Phần 2: Xử lý và Hiển thị kết quả ---
    st.write("---")
    st.subheader("Kết quả AI đọc được:")
    
    # Thêm nút bấm để bắt đầu xử lý (tránh gọi API liên tục mỗi khi load lại app)
    if st.button("🤖 Bắt đầu đọc chữ"):
        # Hiển thị trạng thái đang xử lý
        with st.spinner('AI đang "nhìn" và phân tích chữ viết... vui lòng chờ giây lát...'):
            try:
                # 1. Chuẩn bị prompt kỹ thuật
                prompt = """
                Bạn là một chuyên gia nhận dạng chữ viết tay (OCR).
                Hãy đọc chính xác văn bản trong hình ảnh này.
                Chữ viết có thể nguệch ngoạc, hãy dùng ngữ cảnh tiếng Việt để suy đoán từ chính xác.
                Chỉ trả về văn bản đã đọc được, không thêm bất kỳ diễn giải nào khác.
                """
                
                # 2. Gọi API Gemini Flash
                # Xử lý cả ảnh và prompt cùng lúc
                response = model.generate_content([prompt, image])
                
                # 3. Hiển thị kết quả lên màn hình
                st.success("Tada! AI đã đọc xong:")
                st.text_area(label="", value=response.text, height=200)
                
                # Thêm mẹo cho người dùng
                st.info("💡 Mẹo: Nếu chữ quá xấu hoặc ảnh mờ, AI có thể đọc sai. Hãy chụp ảnh thẳng góc và đủ sáng.")

            except Exception as e:
                st.error(f"Đã xảy ra lỗi khi gọi API. Vui lòng kiểm tra API Key hoặc kết nối mạng. Lỗi chi tiết: {e}")
```
